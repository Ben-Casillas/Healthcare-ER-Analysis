```python
import pandas as pd

pd.set_option('display.max_columns', None)

patients_df = pd.read_csv('./Sample Data/patients.csv')
encounters_df = pd.read_csv('./Sample Data/encounters.csv')

# patients_df.head(10)
encounters_df.head(10)

row_count = len(encounters_df)
encounters_shape = encounters_df.shape

# print(row_count)
# print(encounters_shape)

# Get Data type of all columns
datatypes = patients_df.dtypes
# print(datatypes)


missing_records = encounters_df[encounters_df.isnull().any(axis=1)] # missing any value in any column
# print(len(missing_records))
# missing_records.head()

# missing value in specific columns, CASE MATTERS
missing_passport = patients_df[patients_df['PASSPORT'].isna()]
missing_martial = patients_df[patients_df['MARITAL'].isnull()]
missing_drivers = patients_df[patients_df['DRIVERS'].isnull()]
missing_zip = patients_df[patients_df['ZIP'].isnull() | patients_df['ZIP'] == 0]


married_men = patients_df[(patients_df['MARITAL'] == "M") & (patients_df['GENDER'] == "M")]

#you can use isin() to check if a value is in a list of values
# you also do not need to include the df name within the specified df, you can just use the column name directly
value_in_df = patients_df["CITY"].isin(["New York", "Los Angeles", "Chicago"])

# print for some reason seems to override above df displays
print("Missing Passport: " + str(len(missing_passport)))
print("Missing Marital Status: " + str(len(missing_martial)))
print("Missing Driver's License: " + str(len(missing_drivers)))
print("Missing ZIP Code: " + str(len(missing_zip))) 
print("Married Men: " + str(len(married_men)))
print("People in specified cities: " + str(len(value_in_df)))

#duplicate values
#this reads as "get all rows in patients_df where the BIRTHPLACE column is duplicated, and keep all duplicates"
# you can simply write the name of the dataframe (plus any transformations) to see the output in the console, very cool 
# duplicate will return ALL rows where the specified column is duplicated
encounters_df[encounters_df.duplicated(subset=['PATIENT'], keep=False)].sort_values(by=['PATIENT'])

```
